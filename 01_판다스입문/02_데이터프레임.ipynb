```python
# 행 인덱스 변경
# DataFrame객체.index = 행인덱스배열
# DataFrame객체.columns = 열인덱스배열
df.index = ['학생1', '학생2']
print(df)

df.columns = ['연령','남녀','소속']
print(df)
```

```python
# 행 인덱스 변경 : DataFrame객체.rename(index={기존인덱스: 새인덱스})
# 열 이름 변경 : DataFrame객체.rename(columns={기존이름: 새이름})

import pandas as pd

matrix = [
    [15, '남', '덕영중'],
    [17, '여', '수리중']
]

df = pd.DataFrame(matrix, index=['준서', '예은'], columns=['나이', '성별', '학교'])
df
```

```python
df = df.rename(columns={
    '나이': '연령',
    '성별': '남여',
    '학교': '소속'
    })                      # 요거는 재할당을 꼭 해줘야 함
df
```

```python
df = df.rename(index={
    '준서': '학생1',
    '예은': '학생2'
})                           # 요거도 재할당을 꼭 해줘야 함
df
```

```python
# 행과 열의 위치 바꾸기
# 행, 열 바꾸기 : DataFrame객체.transpose() 또는 T

import pandas as pd

exam_data = {
    '이름': ['서준', '우현', '인아'],
    '수학': [90, 80, 70],
    '영어': [98, 89, 95],
    '음악': [85, 95, 100],
    '체육': [100, 90, 90]
}
df = pd.DataFrame(exam_data)
df

df = df.transpose()
df
```
