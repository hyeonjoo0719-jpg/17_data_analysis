```python
# 배열 속성
arr1 = np.array([1, 2, 3, 4])
arr1.shape   # 튜플 형태로 나옴(1차원 배열이고 4개의 요소를 가지고 있다는 의미)
```

```python
arr2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
arr2.shape
```

```python
arr3 = np.array([
    [[1, 2],
     [3, 4]],
     [[5, 6],
      [7, 8]]
])
arr3.shape
```

```python
# ndim( 몇 차원인지 나타냄)
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)
```

```python
# size( 배열의 총 요소 수를 알려줌)
arr1 = np.array([1, 2, 3, 4])
print(arr1.size)
```

```python
arr2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print(arr2.size)
```

```python
arr3 = np.array([
    [[1, 2],
     [3, 4]],
     [[5, 6],
      [7, 8]]
])
arr3.size
```
