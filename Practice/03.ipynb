## 문제 ① 구급대 평균 이동시간

주어진 데이터를 이용하여 **이동시간**(출동시간과 도착시간의 차이)이  
평균적으로 **가장 오래 걸린 구급대**의 이동시간을 구하시오.

- 분 단위로 환산
- 반올림하여 **정수**로 출력

**데이터**: `data_6_1_1.csv`
```
https://raw.githubusercontent.com/YoungjinBD/dataset/main/data_6_1_1.csv
```

> 출력: 평균 이동시간(분, 정수)

````python
df1 = pd.read_csv("https://raw.githubusercontent.com/YoungjinBD/dataset/main/data_6_1_1.csv")

# 문자열 -> datetime 변환
df1['출동시간'] = pd.to_datetime(df1['출동시간'])
df1['도착시간'] = pd.to_datetime(df1['도착시간'])

# 이동시간(분)
df1['이동시간'] = (df1['도착시간'] - df1['출동시간']).dt.total_seconds() / 60

# 구급대별 평균 이동시간 -> 최댓값
ans = df1.groupby('구급대')['이동시간'].mean().max()
print(round(ans))
````

## 문제 ② 교수 1인당 학생 수가 가장 많은 대학

주어진 데이터를 이용하여 **교수 1인당 맡고 있는 학생 수**가 가장 많은 대학교를 찾고,  
그 대학교의 **전체 교수 수**를 구하시오. (**정수**로 출력)

**데이터**: `data_6_1_2.csv`
```
https://raw.githubusercontent.com/YoungjinBD/dataset/main/data_6_1_2.csv
```

> 출력: 해당 대학교의 전체 교수 수(정수)

````python
df2 = pd.read_csv("https://raw.githubusercontent.com/YoungjinBD/dataset/main/data_6_1_2.csv")

# 학과 컬럼 = 대학교명, 교수를 뺀 나머지
students = df2.drop(columns=['대학교명', '교수']).sum(axis=1)

# 교수 1인당 학생 수
df2['학생수'] = students
df2['비율'] = df2['학생수'] / df2['교수']

# 비율이 가장 큰 대학의 교수 수
idx = df2['비율'].idxmax()
print(df2.loc[idx, '대학교명'], int(df2.loc[idx, '교수']))
````

## 문제 ③ 연도별 총 범죄 건수 월평균

주어진 데이터를 이용하여 **연도별 총 범죄 건수의 월평균** 값을 구하고,  
그 값이 **가장 큰 연도**의 총 범죄 건수 월평균 값을 집계하시오.

- 반올림하여 **정수**로 출력

**데이터**: `data_6_1_3.csv`
```
https://raw.githubusercontent.com/YoungjinBD/dataset/main/data_6_1_3.csv
```

> 출력: 월평균 최댓값(정수)

````python
df3 = pd.read_csv("https://raw.githubusercontent.com/YoungjinBD/dataset/main/data_6_1_3.csv")

# 연도 추출 ('2018년 01월' -> 2018)
df3['연도'] = df3['날짜'].str[:4].astype(int)

# 총 범죄 건수 = 범죄 컬럼 합계 (날짜, 주요검찰청명, 연도 제외)
crime_cols = df3.columns.drop(['날짜', '주요검찰청명', '연도'])
df3['총범죄'] = df3[crime_cols].sum(axis=1)

# 연도별 월평균 -> 최댓값
ans = df3.groupby('연도')['총범죄'].mean().max()
print(round(ans))
````
