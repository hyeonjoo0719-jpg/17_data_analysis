## 문제 ① 음식물쓰레기 2L 종량제 봉투 평균 가격

`trash_bag` 데이터는 지역별 종량제 봉투 가격 정보이다.  
각 용량별 가격 컬럼은 해당 행의 조건을 만족하는 봉투가 존재하면 가격을,  
존재하지 않으면 0을 값으로 갖는다.

용도가 **'음식물쓰레기'**이고 사용 대상이 **'가정용'**인 2L 봉투 가격의  
**평균**을 구하시오. (소수점 이하는 **절사**하여 **정수**로 출력)

**데이터**: `trash_bag.csv`
```
https://raw.githubusercontent.com/YoungjinBD/dataset/main/trash_bag.csv
```

> 출력: 평균 가격(정수, 절사)

````python
df1 = pd.read_csv(
    "https://raw.githubusercontent.com/YoungjinBD/dataset/main/trash_bag.csv",
    encoding="cp949",
)
df1.head()

cond = (df1['용도'] == '음식물쓰레기') & (df1['사용대상'] == '가정용')
target = df1.loc[cond, '2L가격']
target = target[target > 0]          # ← 존재하지 않는 봉투(0) 제외

print(int(target.mean()))            # int()가 곧 절사
````

## 문제 ② BMI 정상·과체중 인원 수 차이

BMI(Body Mass Index)는 몸무게(kg)를 키(m)의 제곱으로 나누어 계산하며,  
비만도를 나타내는 지표이다. 아래 기준에 따라 분류한다.

| BMI 범위 | 분류 |
|---|---|
| 18.5 미만 | 저체중 |
| 18.5 이상 ~ 23 미만 | 정상 |
| 23 이상 ~ 25 미만 | 과체중 |
| 25 이상 ~ 30 미만 | 경도비만 |
| 30 이상 | 중등도비만 |

제공된 BMI 데이터를 기준으로 **정상** 체중 인원과 **과체중** 인원의  
차이를 정수로 출력하시오.

**데이터**: `BMI.csv`
```
https://raw.githubusercontent.com/YoungjinBD/dataset/main/BMI.csv
```

> 출력: (정상 인원 수) − (과체중 인원 수) (정수)

````python


df2 = pd.read_csv("https://raw.githubusercontent.com/YoungjinBD/dataset/main/BMI.csv")
df2.head()

bmi = df2['Weight'] / (df2['Height'] / 100) ** 2

normal = ((bmi >= 18.5) & (bmi < 23)).sum()
over   = ((bmi >= 23)   & (bmi < 25)).sum()

print(abs(normal - over))
````

## 문제 ③ 순전입학생 수가 가장 많은 학교

`students` 데이터는 학교별·학년별 **총 전입학생**, **총 전출학생**,  
**전체학생 수** 정보이다.

순전입학생 수는 다음과 같이 정의한다.

$$\text{순전입학생 수} = \text{총 전입학생} - \text{총 전출학생}$$

**순전입학생 수가 가장 많은 학교**의 **전체학생 수**를 구하시오.

**데이터**: `students.csv`
```
https://raw.githubusercontent.com/YoungjinBD/dataset/main/students.csv
```

> 출력: 해당 학교의 전체학생 수

````python


df3 = pd.read_csv(
    "https://raw.githubusercontent.com/YoungjinBD/dataset/main/students.csv",
    encoding="cp949",
)
df3.head()

print(df3.columns.tolist())
df3.head()

cols = df3.columns

school = [c for c in cols if '학교' in c][0]
in_col  = [c for c in cols if '전입' in c][0]
out_col = [c for c in cols if '전출' in c][0]
total   = [c for c in cols if '전체' in c][0]

print(school, in_col, out_col, total)   # 뭘 잡았는지 확인

g = df3.groupby(school)[[in_col, out_col, total]].sum()
g['순전입'] = g[in_col] - g[out_col]

print(g.loc[g['순전입'].idxmax(), total])
````
