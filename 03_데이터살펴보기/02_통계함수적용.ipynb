```python
# 모든 열의 평균값: DataFrame 객체.mean()
# 특정 열의 평균값: DataFrame 객체['열 이름'].mean()

import pandas as pd
df = pd.read_csv('../data/auto-mpg.csv', header=None)

df.columns = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight',
            'acceleration', 'model year', 'origin', 'name']

print(df.mean(numeric_only=True))
print('\n')
print(df['mpg'].mean())
print(df.mpg.mean())
print('\n')
print(df[['mpg', 'weight']].mean())
```

```python
# 중앙값
df.median(numeric_only=True)
df.mpg.median()
```

```python
df.max()
df.min()
```

```python
df.max(numeric_only=True)  # 숫자 중 가장 큰 값
```

```python
df.min(numeric_only=True)        # 숫자 중 가장 작은 값
```

```python
df.std(numeric_only = True)
```

```python
# 모든 열의 상관계수 : DataFrame 객체.corr()
# 특정 열의 상관계수 : DaTaFrame 객체[열 이름의 리스트].corr()

df.corr(numeric_only=True)   # 상관계수
df[['mpg', 'weight']].corr()
```
