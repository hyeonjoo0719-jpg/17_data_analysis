```python
# 중복 데이터 확인

import pandas as pd
df = pd.DataFrame({'c1':['a', 'a', 'b', 'a', 'b'],
                    'c2':[1, 1, 1, 2, 2],
                    'c3':[1, 1, 2, 2, 2]})
df
```

```python
# 데이터프레임 전체 행 데이터 중에서 중복값 찾기
df_dup_first = df.duplicated()
df_dup_first
```

```python
# 데이터프레임 전체 행 데이터 중에서 중복값 찾기
df_dup_false = df.duplicated(keep=False)
df_dup_false
```

```python
# 데이터프레임의 특정 열 데이터에서 중복값 찾기
col_dup = df['c2'].duplicated()
col_dup
```

```python
# 데이터프레임의 특정 열 데이터에서 중복값 찾기
col_dup2 = df.duplicated(subset=['c2'])
col_dup2
```

```python
# 중복 데이터 제거

import pandas as pd
df = pd.DataFrame({'c1':['a', 'a', 'b', 'a', 'b'],
                    'c2':[1, 1, 1, 2, 2],
                    'c3':[1, 1, 2, 2, 2]})

print(df)

# 중복 행 제거
df2 = df.drop_duplicates()
print(df2)

# 중복 행 제거
df3 = df.drop_duplicates(keep='last')
print(df3)

# 중복 행 제거
df4 = df.drop_duplicates(keep=False)
print(df4)

# c2, c3 열을 기준으로 중복 행 제거
df5 = df.drop_duplicates(subset=['c2', 'c3'])
print(df5)

# c2, c3 열을 기준으로 중복 행 제거(중복 데이터 모두 제거)
df6 = df.drop_duplicates(subset=['c2', 'c3'], keep=False)
print(df6)
```
